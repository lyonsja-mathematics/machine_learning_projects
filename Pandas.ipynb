```python
# Import a data frame from a CSV file
trolley=pd.read_csv("TrolleyGAR.csv")
trolley.head()
```

```python
# Rename columns
trolley=trolley.rename(columns={"Target.Date":"date", "Hospital.Code":"provider", 
                                "Total.8am":"value8am","Total.2pm":'value2pm',
                               'Total.8pm':'value8pm'})
```

```python
# Replace values
trolley.provider=trolley.provider.replace([101,108,500],[1,2,3])
```

```python
# Find rows with NaN values in date column.
trolley[trolley.date.isna()]
```

```python
# Find any rows with NaN values.
trolley[trolley.isna().any(axis=1)]
```

```python
# Remove rows with NaN values in date column.
trolley=trolley.dropna(axis=0, subset=["date"])
trolley.shape
```

```python
# Replace NaNs in value column with 0.
trolley.value8am=trolley.value8am.fillna(0)
```

```python
# Change a column data type
trolley["value8am"]=trolley["value8am"].astype(float)
trolley.provider=trolley.provider.astype(int)
```

```python
# Find duplicated rows
trolley[trolley.duplicated()]
```

```python
# Remove duplicated rows
trolley=trolley.drop_duplicates()
trolley.shape
```

```python
# Create a table of the character lengths of a str type column.
pd.DataFrame(trolley.date.astype(str).str.len().value_counts()).rename_axis('str_len')
```

```python
# Create a new column for year, month and day.
trolley["year"]=trolley.date.dt.year
trolley["month"]=trolley.date.dt.month
trolley['day']=trolley.date.dt.day
trolley.sample(3)
```

```python
# Use the apply method and lambda function to create a if-else column
trolley['season'] = trolley.month.apply(lambda x: 'winter' if x in [1,2,12] else
                                        'spring' if x in [3,4,5] else 
                                        'summer' if x in [6,7,8] else 'autumn')

trolley.sample(3)
```

```python
# Group by two columns and aggregate sum over a third numeric column.
agg_df=trolley[['provider','year','month','value8am']].groupby(
    by=['provider','year','month'], as_index=False).agg({'value8am':['sum']})
agg_df.columns=['provider','year','month','total_value8am']
agg_df.head()
```

```python
# Calculate a running total for each provider and year.
agg_df['running_total']=agg_df.groupby(by=['provider','year']).total_value8am.cumsum()
agg_df.head()
```

```python
# Pivot table
pivotDF=agg_df.query('provider == 923').pivot(index='month', 
                                              columns='year', values='total_value8am').fillna(0)

for i in pivotDF.columns:
    pivotDF[i] = pivotDF[i].astype(int)

pivotDF
```

```python
# Group by two columns and aggregate calculation.
trolley.query("year == 2019").groupby(by=["provider","month"], as_index=False).agg(
    {"value8am":["mean",'sum']})[0:5]
```

```python
# Calculate the 30 day moving average.
trolley['30DMA']=0
for i in range(len(trolley)):
    provider_i=trolley.loc[i, ['provider']][0]
    date_i=trolley.loc[i, ['date']][0]
    last30Days=pd.date_range(end=date_i, periods=30)
    moving_average_i=trolley[
    (trolley.date.isin(last30Days)) & 
    (trolley.provider==provider_i)].value8am.mean()
    trolley.loc[i, '30DMA']=moving_average_i

trolley.tail()
```

```python
# Import CSV data from a directory
path1='C:/Users/C10381149/OneDrive - Technological University Dublin/Documents/Python/'
path2='Datasets/PET_Data'
os.chdir(path1+path2)
pet=pd.DataFrame({'provider':[],'mrn':[],'reg':[],'age':[],
                  'triage':[],'edclinician':[],'depart':[],'destination':[]})
for f in os.listdir():
    pet=pd.concat([pet, pd.read_csv(f, header=0, names=pet.columns)])

pet.sample(3)
```

```python
pet=pet[(pet.reg.astype(str).str.len()==20) & (pet.age.isin(range(0,121)))]
pet.head(3)
```

```python
# Change timestamp columns from character strings to datetime objects
for i in ['reg','triage','edclinician','depart']:
    pet[i] = pd.to_datetime(pet[i], format="%d-%b-%Y %H:%M:%S")    
```

```python
# Patient experience time
pet['pet'] = (pet.depart - pet.reg)/pd.Timedelta(hours=1)
```

```python
# Change the date format
pet['week_day']=pet.reg.dt.strftime('%A')
pet[['reg','week_day']].head()
```

```python
# Merge two data frames on a date column.
meanPET=pet.groupby(by=['provider','depart'], as_index=False).aggregate({'pet':'mean'})
meanPET.columns=['provider','date','meanPET']
mergedDF=pd.merge(left=meanPET, right=trolley[['provider','date','value8am']],
                  left_on=['provider','date'], right_on=['provider','date'],
                  how='inner').sort_values(['provider','date'])
mergedDF.head()
```
